```python
import polars as pl

# 1. Caminho do arquivo Parquet
arquivo_parquet = r"C:\Users\MICRO\OneDrive\Documentos\UC2\contratacao_regioes.parquet"

# 2. Criando a consulta corrigida com "Região"
query_regiao = (
    pl.scan_parquet(arquivo_parquet)
    .group_by("Região")
    .agg(pl.len().alias("total_contratacoes"))
    .sort("total_contratacoes", descending=True)
)

#Regiões com maior contratação

# 3. Executando a consulta e exibindo o resultado
print(query_regiao.collect())

```

```python
import polars as pl

# 1. Caminho do arquivo Parquet
arquivo_parquet = r"C:\Users\MICRO\OneDrive\Documentos\UC2\contratacao_regioes.parquet"

# 2. Criando a consulta ordenada do MENOR para o MAIOR (descending=False)
query_menor_regiao = (
    pl.scan_parquet(arquivo_parquet)
    .group_by("Região")
    .agg(pl.len().alias("total_contratacoes"))
    .sort("total_contratacoes", descending=False) # Ordem crescente
)

# Regiões que teve menor contratações 

# 3. Executando e exibindo o resultado
print(query_menor_regiao.collect())
```
