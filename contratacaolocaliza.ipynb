```python
import polars as pl

# 1. Caminhos dos arquivos
arquivo_parquet = r"C:\Users\MICRO\OneDrive\Documentos\UC2\contratacao_localizar.parquet"
arquivo_excel_cargos = r"C:\Users\MICRO\OneDrive\Documentos\UC2\ranking_cargos_localiza.xlsx"

# 2. Carrega a consulta do arquivo Parquet
stats_query = pl.scan_parquet(arquivo_parquet)

# 3. Cria a tabela de cargos que mais contrataram (do maior para o menor)
tabela_cargos = (
    stats_query.group_by("Cargo")
    .agg(pl.len().alias("Total_Contratacoes"))
    .sort("Total_Contratacoes", descending=True)
)

# 4. Executa a consulta
df_resultado = tabela_cargos.collect()

# Cargos que mas teve contratações
print(df_resultado)

# 6. Salva o resultado em um arquivo Excel pronto
df_resultado.write_excel(arquivo_excel_cargos)
print(f"\nArquivo Excel gerado com sucesso em: {arquivo_excel_cargos}")
```
