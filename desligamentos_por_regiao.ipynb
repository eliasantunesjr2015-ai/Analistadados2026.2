```python
import polars as pl

# 1. Caminhos corrigidos (Origem é o arquivo que já existe; Saída é o novo que vamos criar)
arquivo_origem = r"C:\Users\MICRO\OneDrive\Documentos\UC2\desligamento_faixaetaria.parquet"
arquivo_saida = r"C:\Users\MICRO\OneDrive\Documentos\UC2\desligamentos_por_regiao.parquet"

# 2. Processando os dados usando o arquivo de ORIGEM correto
df_regiao = (
    pl.scan_parquet(arquivo_origem)  # <-- Corrigido aqui para ler o arquivo que existe
    .group_by("Região")
    .agg(
        pl.len().alias("Total_Desligamentos")
    )
    .sort("Total_Desligamentos", descending=True)
    .collect()
)

# 3. Salvando o resultado no arquivo de SAÍDA
df_regiao.write_parquet(arquivo_saida)
print(f"✅ Arquivo 'desligamentos_por_regiao.parquet' gerado com sucesso!")

# 4. Configuração para exibir todas as linhas na tela
with pl.Config(tbl_rows=10000):
    print("\n--- DADOS SALVOS NO ARQUIVO ---")
    print(df_regiao)

```

```python
import polars as pl

# 1. Caminho do arquivo original de desligamentos
arquivo_origem = r"C:\Users\MICRO\OneDrive\Documentos\UC2\desligamento_faixaetaria.parquet"

# 2. Criar a consulta agrupando por Região e trazendo o menor volume para o topo
query_regiao_lanterna = (
    pl.scan_parquet(arquivo_origem)
    .group_by("Região")
    .agg(
        pl.len().alias("Total_Desligamentos")
    )
    # descending=False organiza do MENOR para o MAIOR número de desligamentos
    .sort("Total_Desligamentos", descending=False)
)

# 3. Executar e exibir apenas o primeiro colocado da lista invertida
resultado = query_regiao_lanterna.limit(1).collect()

print("\n--- REGIÃO COM MENOS DESLIGAMENTOS ---")
print(resultado)

```
