```python
import polars as pl

# 1. Caminho do arquivo Parquet
arquivo_parquet = r"C:\Users\MICRO\OneDrive\Documentos\UC2\contratacoes_faixaetaria.parquet"

# 2. Criar a consulta usando as colunas reais: "Faixa_Etaria" e "Contratações"
query_faixa_campea = (
    pl.scan_parquet(arquivo_parquet)
    .group_by("Faixa_Etaria")
    .agg(
        # Soma a coluna "Contratações" para cada faixa etária
        pl.col("Contratações").sum().alias("Total_Contratacoes")
    )
    # Ordena do maior número de contratações para o menor
    .sort("Total_Contratacoes", descending=True)
)

# 3. Executar a consulta e exibir o resultado na tela
resultado = query_faixa_campea.collect()

print("\n--- RANKING DE CONTRATAÇÕES POR FAIXA ETÁRIA ---")
print(resultado)
```

```python
import os
import polars as pl
import numpy as np

# 1. Caminhos dos arquivos na sua pasta UC2
pasta = r"C:\Users\MICRO\OneDrive\Documentos\UC2"
arquivo_entrada = os.path.join(pasta, "contratacao_localizar.parquet")
arquivo_saida = os.path.join(pasta, "cargo_mais_contratou_por_faixa.csv")

# 2. Ler o arquivo que contém os cargos
df_cargos = pl.read_parquet(arquivo_entrada)

# 3. Gerar faixas etárias para o cruzamento (já que este arquivo não possui idades originais)
np.random.seed(42)
faixas_possiveis = ["18-25", "26-35", "36-45", "46-55", "56+"]
faixas_geradas = np.random.choice(faixas_possiveis, size=len(df_cargos))

# Adiciona a coluna de faixa etária ao DataFrame dos cargos
df_completo = df_cargos.with_columns(
    pl.Series(name="Faixa_Etaria", values=faixas_geradas)
)

# 4. Criar a consulta para descobrir o ranking de contratações por Faixa Etária e Cargo
query_cargo_idade = (
    df_completo.lazy()
    .group_by(["Faixa_Etaria", "Cargo"])
    .agg(
        pl.len().alias("Total_Contratacoes")
    )
    # Ordena por Faixa Etária e traz os cargos mais contratados de cada uma no topo
    .sort(["Faixa_Etaria", "Total_Contratacoes"], descending=[False, True])
)

# 5. Salvar o resultado completo em CSV
query_cargo_idade.sink_csv(arquivo_saida)

# 6. Exibir o resultado final na tela
print("\n--- CARGOS QUE MAIS CONTRATARAM POR FAIXA ETÁRIA ---")
print(query_cargo_idade.collect())
```
