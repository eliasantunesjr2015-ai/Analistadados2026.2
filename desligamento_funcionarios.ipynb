```python
import polars as pl

# 1. Caminhos dos arquivos
arquivo_csv = r"C:\Users\MICRO\OneDrive\Documentos\UC2\desligamento_funcionarios.csv"
arquivo_parquet = r"C:\Users\MICRO\OneDrive\Documentos\UC2\desligamento_funcionarios.parquet"

# 2. Lendo o CSV informando que o separador é ponto e vírgula
df = pl.read_csv(arquivo_csv, separator=";")

# 3. Salvando o arquivo completo no formato Parquet
df.write_parquet(arquivo_parquet)

# 4. Configuração para ler o Parquet e calcular o total de desligamentos por gênero
with pl.Config(tbl_rows=10000):
    # Carrega a consulta do Parquet de forma otimizada (Lazy)
    stats_query = pl.scan_parquet(arquivo_parquet)
    
    # Agrupa pela coluna de Gênero e conta o total de registros
    # NOTA: Certifique-se de que o nome da coluna no seu CSV é exatamente 'Genero' (ou ajuste abaixo)
    resultado = (
        stats_query
        .group_by("Genero")
        .len(name="Total_Desligamentos")
    )
    
    # Executa a consulta e mostra o resultado na tela
    print(resultado.collect())
```

```python
import polars as pl

# 1. Caminhos dos arquivos
arquivo_csv = r"C:\Users\MICRO\OneDrive\Documentos\UC2\desligamento_funcionarios.csv"
arquivo_parquet = r"C:\Users\MICRO\OneDrive\Documentos\UC2\desligamento_funcionarios.parquet"

# 2. Lendo o CSV informando que o separador é ponto e vírgula
df = pl.read_csv(arquivo_csv, separator=";")

# 3. Salvando o arquivo completo no formato Parquet
df.write_parquet(arquivo_parquet)

# 4. Configuração para ler o Parquet e encontrar a função com mais desligamentos
with pl.Config(tbl_rows=10000):
    # Carrega a consulta do Parquet de forma otimizada (Lazy)
    stats_query = pl.scan_parquet(arquivo_parquet)
    
    # Agrupa por Funcao (corrigido), conta o total e ordena do maior para o menor
    resultado = (
        stats_query
        .group_by("Funcao")
        .len(name="Total_Desligamentos")
        .sort("Total_Desligamentos", descending=True)
    )
    
    # Executa a consulta e mostra o ranking na tela
    print(resultado.collect())
```
